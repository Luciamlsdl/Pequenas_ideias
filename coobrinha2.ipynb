```python
%pip install pygame

import pygame
import random

# Configurações do jogo
LARGURA, ALTURA = 800, 600
TAMANHO_QUADRADO = 20
VELOCIDADE_INICIAL = 10

# Cores
PRETO = (0, 0, 0)
BRANCO = (255, 255, 255)
VERDE = (0, 255, 0)
VERMELHO = (255, 0, 0)

# Inicializar Pygame
pygame.init()
tela = pygame.display.set_mode((LARGURA, ALTURA))
pygame.display.set_caption("Cobrinha - Python")
relogio = pygame.time.Clock()

# Funções
def desenhar_cobra(corpo):
    for segmento in corpo:
        pygame.draw.rect(tela, BRANCO, [segmento[0], segmento[1], TAMANHO_QUADRADO, TAMANHO_QUADRADO])

def gerar_comida():
    x = random.randint(0, (LARGURA - TAMANHO_QUADRADO) // TAMANHO_QUADRADO) * TAMANHO_QUADRADO
    y = random.randint(0, (ALTURA - TAMANHO_QUADRADO) // TAMANHO_QUADRADO) * TAMANHO_QUADRADO
    return x, y

# Função principal do jogo
def rodar_jogo():
    x = LARGURA // 2
    y = ALTURA // 2
    velocidade_x, velocidade_y = TAMANHO_QUADRADO, 0
    corpo_cobra = [[x, y]]
    tamanho_cobra = 3
    pontuacao = 0
    comida = gerar_comida()
    rodando = True

    while rodando:
        tela.fill(PRETO)
        for evento in pygame.event.get():
            if evento.type == pygame.QUIT:
                rodando = False
            elif evento.type == pygame.KEYDOWN:
                if evento.key == pygame.K_UP and velocidade_y == 0:
                    velocidade_x, velocidade_y = 0, -TAMANHO_QUADRADO
                elif evento.key == pygame.K_DOWN and velocidade_y == 0:
                    velocidade_x, velocidade_y = 0, TAMANHO_QUADRADO
                elif evento.key == pygame.K_LEFT and velocidade_x == 0:
                    velocidade_x, velocidade_y = -TAMANHO_QUADRADO, 0
                elif evento.key == pygame.K_RIGHT and velocidade_x == 0:
                    velocidade_x, velocidade_y = TAMANHO_QUADRADO, 0

        # Atualiza a posição da cobra
        x += velocidade_x
        y += velocidade_y
        corpo_cobra.append([x, y])
        if len(corpo_cobra) > tamanho_cobra:
            del corpo_cobra[0]

        # Colisão com as bordas ou com o próprio corpo
        if x < 0 or x >= LARGURA or y < 0 or y >= ALTURA or [x, y] in corpo_cobra[:-1]:
            rodando = False

        # Comer comida
        if x == comida[0] and y == comida[1]:
            tamanho_cobra += 1
            pontuacao += 10
            comida = gerar_comida()

        # Desenhar elementos
        desenhar_cobra(corpo_cobra)
        pygame.draw.rect(tela, VERDE, [comida[0], comida[1], TAMANHO_QUADRADO, TAMANHO_QUADRADO])
        fonte = pygame.font.SysFont(None, 35)
        texto = fonte.render(f"Pontos: {pontuacao}", True, VERMELHO)
        tela.blit(texto, (10, 10))

        pygame.display.update()
        relogio.tick(VELOCIDADE_INICIAL)

    pygame.quit()

# Executar o jogo
rodar_jogo()
```
